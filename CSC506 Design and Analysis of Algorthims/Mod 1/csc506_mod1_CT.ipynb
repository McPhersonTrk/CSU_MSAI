```python
# Search for a target item using a linear search algorithm
def Linear_serch(items, target):
    found_index =[]
    for index, item in enumerate(items):
        if item == target:
            found_index.append(index)
    return found_index
# items  in the marketplace
items = ['kitten', 'puppy', 'rabbit', 'bird', 'turtle', 'hamster', 'fish', 'frog',
        'snake', 'lizard', 'parrot', 'cat', 'dog', 'guinea pig', 'canary', 'hamster',
        'mouse', 'gerbil', 'goldfish', 'ferret', 'chinchilla', 'hedgehog', 'tarantula',
        'sugar glider', 'chameleon']

#Search for a target item in the marketplace
while True:
    target= input("Enter your query: (Press q to quit)").strip().lower()
    if target == 'exit':
        print("Goodbye")
        break
# Perform a linear search for the target in the items
    found_index = Linear_serch(items, target)
    if found_index:
        print(f"Iten found at index: {', '.join(map(str, found_index))}")
    else:
        print(f"Item '{target}' not found in the database.")
# user input continue or quit
    continue_search = input("Do you want to search for another item? (yes/no): ").lower()
    if continue_search != 'yes':
        print("Thank you for using the search. Goodbye!")
        break

```
