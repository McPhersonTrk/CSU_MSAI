```python
# COntent Recomendation system using a Hash Table
# CRS = content_recommendation_system

# Initialize the class for the COntent recomendation System.
class CRS:
    def __init__(self, capacity=1000):
        self.capacity = capacity
        self.table = [None] * self.capacity
    # define a hash() function that generates a hash index for a given key.
    def _hash_function(self, key):
        try:
            return hash(key) % self.capacity
        except Exception as e:
            raise ValueError(f"Error computing hash for key {key}: {e}")
    # define a function that adds content to the table, associating it with a user ID.
    # The function takes in a user ID and a content ID as arguments.
    def content_add(self, user_id, content_id):
        if not isinstance(user_id, str) or not isinstance(content_id, str):
            raise TypeError("Both user_id and content_id must be strings.")
        
        index = self._hash_function(user_id)
        if not self.table[index]:
            self.table[index] = []

        if content_id not in self.table[index]:
            self.table[index].append(content_id)
        else:
            print(f"Content ID {content_id} is already associated with user ID {user_id}.")
    # Define a function that retrieves recommendations for a specific user ID.
    # The function takes in a user ID as an argument and returns a list of content IDs associated with the user ID.
    def get_recommendations(self, user_id):
        
        if not isinstance(user_id, str):
            raise TypeError("user_id must be a string.")

        index = self._hash_function(user_id)
        if self.table[index] is None:
            print(f"No recommendations found for user ID {user_id}.")
            return []

        return self.table[index]

# Sample usage of the system
if __name__ == "__main__":
    system = CRS(capacity=1024)

    try:
        # Simulate adding content for different users
        system.content_add("user123", "content456")
        system.content_add("user123", "content789")
        system.content_add("user456", "content123")
        system.content_add("user123", "content456")  

        # Retrieve recommendations for a user
        user_recommendations = system.get_recommendations("user123")
        print(f"Recommendations for user123: {user_recommendations}")

        # Retrieve recommendations for another user
        another_user_recommendations = system.get_recommendations("user456")
        print(f"Recommendations for user456: {another_user_recommendations}")

        # Test for a user with no recommendations
        empty_recommendations = system.get_recommendations("user789")
        print(f"Recommendations for user789: {empty_recommendations}")

    except Exception as e:
        print(f"An error occurred: {e}")
```

```python
# Content Recommendation System using a Hash Table
# CRS = content_recommendation_system
import csv

# Initialize a class for the Content Recommendation System
class CRS:
    def __init__(self, capacity=1000):
        self.capacity = capacity
        self.table = [None] * self.capacity

    # Define a hash function that generates a hash index for a given key
    def _hash_function(self, key):
        try:
            return hash(key) % self.capacity
        except Exception as e:
            raise ValueError(f"Error computing hash for key {key}: {e}")

    # Add content to the table, associating it with a user ID
    def content_add(self, user_id, content_id):
        if not isinstance(user_id, str) or not isinstance(content_id, str):
            raise TypeError("user_id and content_id must be strings.")

        index = self._hash_function(user_id)
        if not self.table[index]:
            self.table[index] = []

        if content_id not in self.table[index]:
            self.table[index].append(content_id)
        
    # Retrieve recommendations for a specific user ID
    def get_recommendations(self, user_id):
        if not isinstance(user_id, str):
            raise TypeError("user_id must be a string.")

        index = self._hash_function(user_id)
        if not self.table[index]:
            self.table[index] = {}
        #create dictionary for each index to store useer_id to content
        if user_id not in self.table[index]:
            self.table[index][user_id] = []
        if content_id not in self.table[index][user_id]:
            self.table[index][user_id].append(content_id)
        
    # Retrieve all non-empty buckets for verification
    def get_summary(self):
            non_empty_buckets = sum(1 for bucket in self.table if bucket)
            total_users = sum(len(bucket) for bucket in self.table if bucket)
            return non_empty_buckets, total_users
    

    # Retrieve all recommendations for all users
    def get_all_recommendations(self):
        all_recommendations = {}
        for index, bucket in enumerate(self.table):
            if bucket:
                all_recommendations[index] = bucket
        return all_recommendations

# Load CSV data into the CRS system
if __name__ == "__main__":
    data_path = r"C:\Users\mikea\Desktop\CSU-MSAI\CSU_MSAI_GitHub\CSC506 Design and Analysis of Algorthims\Mod 5\social media influencers - instagram sep-2022.csv"
    system = CRS(capacity=1024)

    try:
        # Initialize counters TO TRACK PROCCEssed records
        total_records = 0
        skipped_records = 0
        
        # Load CSV dataset into the CRS program.
        with open(data_path, mode='r', encoding='utf-8') as file:
            csvFile = csv.reader(file)
            next(csvFile)  # Skip the header row
            for lines in csvFile:
                # Use "Instagram name" as the user ID
                user_id = lines[1].strip()  # Instagram name
                # Combine Category_1 and Category_2 as the content ID for each user
                content_id = ",".join(filter(None, [lines[7].strip(), lines[8].strip()]))
                if not content_id:  # Skip users with no valid categories
                    skipped_records += 1
                    continue
                system.content_add(user_id, content_id)
                total_records += 1
        # Get summary of the hash table
        non_empty_buckets, total_users = system.get_summary()
        print(f"Total records processed: {total_records}")
        print(f"Records skipped due to missing categories: {skipped_records}")
        print(f"Non-empty buckets in the hash table: {non_empty_buckets}")
        print(f"Total user-content associations stored: {total_users}")

        # retrieve and save recomendations for all users
        all_recomendations = {}
        for index, bucket in enumerate(system.table):
            if bucket:
                for user in bucket:
                    user_recomendations[user] = system.get_recommendations(user[0])
                    all_recomendations[user[0]] = user_recomendations 

        # save the recomendations as a CSV file
        output_file = "user_recommendations.csv"
        with open(output_file, mode='w', newline='', encoding='utf-8') as out_file:
            writer = csv.writer(out_file)
            writer.writerow(["User ID", "Recommendations"])
            for user_id, recommendations in all_recommendations.items():
                writer.writerow([user_id, "; ".join(recommendations)])

        print(f"All user recommendations have been saved to {output_file}")

    except Exception as e:
        print(f"An error occurred: {e}")
```

```python
# Content Recommendation System using a Hash Table
# CRS = content_recommendation_system
import csv

# Initialize the class for the Content Recommendation System
class CRS:
    def __init__(self, capacity=1000):
        self.capacity = capacity
        self.table = [None] * self.capacity

    # Define a hash function that generates a hash index for a given key
    def _hash_function(self, key):
        try:
            return hash(key) % self.capacity
        except Exception as e:
            raise ValueError(f"Error computing hash for key {key}: {e}")

    # Add content to the table, associating it with a user ID
    def content_add(self, user_id, content_id):
        if not isinstance(user_id, str) or not isinstance(content_id, str):
            raise TypeError("user_id and content_id must be strings.")

        index = self._hash_function(user_id)
        if not self.table[index]:
            self.table[index] = {}

        # Use a dictionary at each index to store user_id -> content
        if user_id not in self.table[index]:
            self.table[index][user_id] = []
        
        if content_id not in self.table[index][user_id]:
            self.table[index][user_id].append(content_id)

    # Retrieve recommendations for a specific user ID
    def get_recommendations(self, user_id):
        if not isinstance(user_id, str):
            raise TypeError("user_id must be a string.")

        index = self._hash_function(user_id)
        if self.table[index] is None or user_id not in self.table[index]:
            return []

        return self.table[index][user_id]
    
    # Retrieve all user recommendations
    def get_all_recommendations(self):
        all_recommendations = {}
        for bucket in self.table:
            if bucket:
                for user_id, recommendations in bucket.items():
                    all_recommendations[user_id] = recommendations
        return all_recommendations

    # Retrieve summary statistics of the hash table
    def get_summary(self):
        non_empty_buckets = sum(1 for bucket in self.table if bucket)
        total_users = sum(len(bucket) for bucket in self.table if bucket)
        return non_empty_buckets, total_users


# Load CSV data into the CRS system and process all records
if __name__ == "__main__":
    data_path = r"C:\Users\mikea\Desktop\CSU-MSAI\CSU_MSAI_GitHub\CSC506 Design and Analysis of Algorthims\Mod 5\social media influencers - instagram sep-2022.csv"
    system = CRS(capacity=1024)

    try:
        # Initialize counters
        total_records = 0
        skipped_records = 0

        # Load CSV data
        with open(data_path, mode='r', encoding='utf-8') as file:
            csvFile = csv.reader(file)
            next(csvFile)  # Skip the header row
            for lines in csvFile:
                # Use "Instagram name" as the user ID
                user_id = lines[1].strip()  # Instagram name
                # Combine Category_1 and Category_2 as the content ID
                content_id = ",".join(filter(None, [lines[7].strip(), lines[8].strip()]))
                if not content_id:  # Skip users with no valid categories
                    skipped_records += 1
                    continue
                system.content_add(user_id, content_id)
                total_records += 1

        # Get summary of the hash table
        non_empty_buckets, total_users = system.get_summary()
        print(f"Total records processed: {total_records}")
        print(f"Records skipped due to missing categories: {skipped_records}")
        print(f"Non-empty buckets in the hash table: {non_empty_buckets}")
        print(f"Total user-content associations stored: {total_users}")

        # Retrieve and save recommendations for all users
        all_recommendations = system.get_all_recommendations()
        output_file = "user_recommendations.csv"
        with open(output_file, mode='w', newline='', encoding='utf-8') as out_file:
            writer = csv.writer(out_file)
            writer.writerow(["User ID", "Recommendations"])
            for user_id, recommendations in all_recommendations.items():
                writer.writerow([user_id, "; ".join(recommendations)])

        print(f"All user recommendations have been saved to {output_file}")

    except Exception as e:
        print(f"An error occurred: {e}")
```
