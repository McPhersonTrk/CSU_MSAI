```python
# Bubble Sorting Algorithm
#   1. Compare the first two elements
#   2. If the first element is greater than the second element, swap them
#   3. Compare the second and third elements
#   4. If the second element is greater than the third element, swap them
#   5. Continue this process until the end of the list
#   6. Repeat the process for the entire list until no more swaps are needed

# The dataset is a synthetic dataset from Kaggle that contains information about patients in a hospital.
# The dataset has 55,501 rows and 15 columns. 
# The columns are as follows:
# Name, Age, Gender, Blood Type, Medical Condition, Date of Admission, Doctor, Hospital, Insurance Provider, Billing Amount,
# Room Number, Admission Type, Discharge Date, Medication, Test Results

# Importing necessary libraries
import time
import csv
import pandas as pd

# Define the CSV file path
path = r'C:\Users\mikea\Desktop\CSU-MSAI\CSU_MSAI_GitHub\CSC506 Design and Analysis of Algorthims\Mod3_Healthcare\healthcare_dataset.csv'

# Function to create a DataFrame from the CSV file
def create_dataframe(file):
    df = pd.read_csv(file)
    return df

# Bubble Sort implementation to sort rows based on a specific column
def bubble_sort_rows(df, column_name):
    rows = df.values.tolist()  # Convert DataFrame rows to a list of lists
    n = len(rows)
    column_index = df.columns.get_loc(column_name)  # Get the index of the sorting column

    for i in range(n):
        swapped = False
        for j in range(n - i - 1):
            if rows[j][column_index] > rows[j + 1][column_index]:
                rows[j], rows[j + 1] = rows[j + 1], rows[j]  # Swap entire rows
                swapped = True
        if not swapped:  # No swaps mean the list is sorted
            break

    sorted_df = pd.DataFrame(rows, columns=df.columns)  # Convert back to DataFrame
    return sorted_df

# Function to measure the execution time of the bubble sort algortihtm
def measure_time(df, column_name):
    start = time.time()
    sorted_df = bubble_sort_rows(df, column_name)
    end = time.time()
    return sorted_df, end - start

# Main execution, This function will call the above functions to sort the DataFrame based on a specific column
if __name__ == "__main__":
    # Load the DataFrame
    df = create_dataframe(path)

 # Print available column names to assist the user
    print("Available columns in the dataset:")
    print(df.columns.tolist())

    # Take user input for the column to sort by
    column_to_sort = input("Enter the column name to sort by: ").strip()

    # Check if the input column exists in the DataFrame
    if column_to_sort not in df.columns:
        print(f"Error: Column '{column_to_sort}' not found in the dataset.")
    else:
        # Perform the sorting and measure the execution time
        sorted_df, execution_time = measure_time(df, column_to_sort)

        # Print the results
        print("Sorted DataFrame (first 5 rows):")
        print(sorted_df.head(5))  # Display the first 5 rows of the sorted DataFrame
        print(f"Time to run Bubble Sort Algorithm: {execution_time:.4f} seconds")
```

```python
# Merge Sorting Algorithm
#   1. Divide the list into two halves
#   2. Recursively sort each half
#   3. Merge the two halves together

import time
import pandas as pd

# Define the CSV file path
path = r'C:\Users\mikea\Desktop\CSU-MSAI\CSU_MSAI_GitHub\CSC506 Design and Analysis of Algorthims\Mod3_Healthcare\healthcare_dataset.csv'

# Function to create a DataFrame from the CSV file
def create_dataframe(file):
    df = pd.read_csv(file)
    return df

# Merge Sort implementation to sort rows based on a specific column
def merge_sort_rows(rows, column_index):
    if len(rows) <= 1:
        return rows

    # Divide the rows into two halves
    mid = len(rows) // 2
    left_half = merge_sort_rows(rows[:mid], column_index)
    right_half = merge_sort_rows(rows[mid:], column_index)

    # Merge the two halves
    return merge(left_half, right_half, column_index)

def merge(left, right, column_index):
    sorted_rows = []
    i = j = 0

    # Merge the two halves while maintaining order
    while i < len(left) and j < len(right):
        if left[i][column_index] <= right[j][column_index]:
            sorted_rows.append(left[i])
            i += 1
        else:
            sorted_rows.append(right[j])
            j += 1

    # Add any remaining rows from the left and right halves
    sorted_rows.extend(left[i:])
    sorted_rows.extend(right[j:])

    return sorted_rows

# Function to sort the DataFrame rows based on a specific column using merge sort
def merge_sort_df(df, column_name):
    rows = df.values.tolist()  # Convert DataFrame rows to a list of lists
    column_index = df.columns.get_loc(column_name)  # Get the index of the sorting column
    sorted_rows = merge_sort_rows(rows, column_index)  # Perform merge sort
    sorted_df = pd.DataFrame(sorted_rows, columns=df.columns)  # Convert back to DataFrame
    return sorted_df

# Function to measure the execution time of the merge sort
def measure_time(df, column_name):
    start = time.time()
    sorted_df = merge_sort_df(df, column_name)
    end = time.time()
    return sorted_df, end - start

# Main execution
if __name__ == "__main__":
    # Load the DataFrame
    df = create_dataframe(path)

# Print available column names to assist the user
    print("Available columns in the dataset:")
    print(df.columns.tolist())

    # Take user input for the column to sort by
    column_to_sort = input("Enter the column name to sort by: ").strip()

    # Check if the input column exists in the DataFrame
    if column_to_sort not in df.columns:
        print(f"Error: Column '{column_to_sort}' not found in the dataset.")
    else:
        # Perform the sorting and measure the execution time
        sorted_df, execution_time = measure_time(df, column_to_sort)

        # Print the results
        print("Sorted DataFrame (first 5 rows):")
        print(sorted_df.head(5))  # Display the first 5 rows of the sorted DataFrame
        print(f"Time to run Bubble Sort Algorithm: {execution_time:.4f} seconds")

    
```
